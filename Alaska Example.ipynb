```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic Alaska stock price series
np.random.seed(42)
dates = pd.date_range(end=pd.Timestamp('2025-04-20'), periods=365, freq='D')
prices = np.random.normal(loc=50, scale=5, size=len(dates)).cumsum()

# Create DataFrame
alaska_stock = pd.DataFrame({'Date': dates, 'Price': prices})

# Filter last 12 months
last_12_months = alaska_stock[alaska_stock['Date'] >= pd.Timestamp('2024-04-20')]

# Show simple statistics
display(last_12_months.describe())

# Plot the stock prices over the last 12 months
plt.figure(figsize=(10, 5))
plt.plot(last_12_months['Date'], last_12_months['Price'], label='Alaska Stock Price')
plt.xlabel('Date')
plt.ylabel('Price')
plt.title('Alaska Stock Prices Over Last 12 Months')
plt.legend()
plt.grid(True)
plt.show()# This is a test
```
